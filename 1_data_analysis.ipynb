```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from collections import Counter
```

```python
pd.set_option('display.max_columns', None)
```

```python
df = pd.read_parquet("/Users/aniket/TU_Eindhoven/2_Study/Q4_2AMV10_Visual_Analytics/4_Code/2_AMV10_Visual_Analytics/processed_data/movies_data_updated.parquet")
df.sample(5)
```

```python
genre_counts = Counter(genre for genres in df['genre_list'] for genre in genres if isinstance(genres, list))
genre_df = pd.DataFrame(genre_counts.items(), columns=["Genre", "Count"]).sort_values(by="Count", ascending=False)

# Plot
fig = px.bar(genre_df.head(15), x="Genre", y="Count", title="Top 15 Genres by Movie Count")
fig.show()
```

```python
avg_rating_by_genre = df.groupby("main_genre")["vote_average"].mean().reset_index()
avg_rating_by_genre = avg_rating_by_genre.sort_values(by="vote_average", ascending=False)

fig = px.bar(avg_rating_by_genre, x="main_genre", y="vote_average", 
             title="Average Vote by Main Genre", labels={"main_genre": "Genre", "vote_average": "Avg Rating"})
fig.show()
```

```python
# country_counts = df['main_country'].value_counts().reset_index()
# country_counts.columns = ["Country", "Count"]

# fig = px.choropleth(country_counts, locations="Country", locationmode="country names",
#                     color="Count", title="Number of Movies by Country", color_continuous_scale="Viridis")
# fig.show()

country_counts = df['main_country'].value_counts().reset_index()
country_counts.columns = ["Country", "Count"]

fig = px.choropleth(
    country_counts,
    locations="Country",
    locationmode="country names",
    color="Count",
    title="Number of Movies by Country",
    color_continuous_scale="Viridis",
    projection="natural earth"  # You can try 'equirectangular', 'orthographic', etc.
)

# Update layout for size and spacing
fig.update_layout(
    width=1000,
    height=600,
    geo=dict(
        showframe=False,
        showcoastlines=True,
        projection_type='natural earth'
    ),
    margin=dict(l=0, r=0, t=50, b=0)
)

fig.show()
```

```python
df['budget'] = pd.to_numeric(df['budget'], errors='coerce')
df['revenue'] = pd.to_numeric(df['revenue'], errors='coerce')
df['popularity_score'] = pd.to_numeric(df['popularity_score'], errors='coerce')

# Filter valid entries
filtered_df = df[
    df['budget'].notna() & (df['budget'] > 0) &
    df['revenue'].notna() & (df['revenue'] > 0) &
    df['popularity_score'].notna()
]

fig = px.scatter(
    filtered_df,
    x="budget", y="revenue",
    size="popularity_score", color="main_genre",
    hover_data=["title", "vote_average", "main_language"],
    log_x=True, log_y=True,
    title="Revenue vs Budget (Log Scale)"
)
fig.show()
```

```python
year_genre = df.groupby(['release_year', 'main_genre']).size().reset_index(name='count')
fig = px.line(year_genre, x='release_year', y='count', color='main_genre',
              title="Movies Released per Year by Genre", markers=True)
fig.show()
```
