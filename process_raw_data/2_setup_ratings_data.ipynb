```python
import pandas as pd
```

```python
ratings_file_path = '/Users/aniket/.cache/kagglehub/datasets/grouplens/movielens-latest-full/versions/1/ratings.csv'
df_ratings = pd.read_csv(
    ratings_file_path,
    # header=None, # Indicate that there is no header row in the file
    dtype={'userId': 'int32', 'movieId': 'int32', 'rating': 'float32', 'timestamp': 'int64'}
    # If you don't need the timestamp initially, you could drop it here using usecols=['userId', 'movieId', 'rating']
)
```

```python
df_movies = pd.read_parquet("/Users/aniket/TU_Eindhoven/2_Study/Q4_2AMV10_Visual_Analytics/4_Code/2_AMV10_Visual_Analytics/processed_data/movies_data_updated.parquet")
```

```python
df_movies
```

```python
movie_cols_to_merge = ['movieId', 'title', 'main_genre', 'director', 'lead_actor', 'release_year']
df_movies_subset = df_movies[movie_cols_to_merge]
```

```python
df_ratings_with_details = pd.merge(
        df_ratings,
        df_movies_subset, # Merge with the subset of movie details
        on='movieId',
        how='inner' # Only keep ratings for movies that have details
    )
df_ratings_with_details
```

```python
df_ratings_with_details["timestamp"] = pd.to_datetime(df_ratings_with_details["timestamp"], unit = "s")
```
